```python
def main():
    # Get the year to check from the user
    year = int(input('Please input a year: '))

    if year % 4 == 0:  # Checking whether the provided year is evenly divisibly by 4
        if year % 100 == 0:  # Checking whether the provided year is evenly divisibly by 100
            if year % 400 == 0:  # Checking whether the provided year is evenly divisibly by 400
                print("That's a leap year!")
            else:  # (Not divisible by 400)
                print("That's not a leap year.")
        else:  # (Not divisible by 100)
            print("That's a leap year!")
    else:  # (Not divisible by 4)
        print("That's not a leap year.")


# There is no need to edit code beyond this point

if __name__ == '__main__':
    main()
```

```python
import random
from typing import List

N_NUMBERS: int = 10
MIN_VALUE: int = 1
MAX_VALUE: int = 100

def generate_random_numbers(count: int, min_val: int, max_val: int) -> List[int]:
    """Generates a list of random numbers (pure function)."""
    return [random.randint(min_val, max_val) for _ in range(count)]

def process_numbers(numbers: List[int]) -> dict:
    """Processes numbers and returns statistics (pure function)."""
    return {
        'numbers': numbers,
        'total': sum(numbers),
        'average': sum(numbers) / len(numbers) if numbers else 0,
        'maximum': max(numbers) if numbers else None,
        'minimum': min(numbers) if numbers else None,
        'sorted_asc': sorted(numbers),
        'sorted_desc': sorted(numbers, reverse=True)
    }

def display_results(stats: dict) -> None:
    """Displays the results (side effect only in this function)."""
    print("Generated Numbers:", stats['numbers'])
    print("Total:", stats['total'])
    print("Average:", stats['average'])
    print("Max Value:", stats['maximum'])
    print("Min Value:", stats['minimum'])
    print("Sorted (Ascending):", stats['sorted_asc'])
    print("Sorted (Descending):", stats['sorted_desc'])

def main():
    """Main function that orchestrates the program flow."""
    numbers = generate_random_numbers(N_NUMBERS, MIN_VALUE, MAX_VALUE)
    stats = process_numbers(numbers)
    display_results(stats)

if __name__ == '__main__':
    main()
```
